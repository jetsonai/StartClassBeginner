## S14. 변수로 계산하기

## S19. 장바구니 만들기와 인덱싱
리스트는 대괄호 `[]`로 만들고, 번호(인덱스)는 **0부터** 셉니다. `[-1]`은 뒤에서 첫 번째예요.

````python
cart = ["아메리카노", "카페라테", "치즈케이크"]
print(cart[0])
print(cart[2])
print(cart[-1])
print(len(cart))
````

## S21. 일부만 보기, 들어 있는지 확인하기
슬라이싱 `[시작:끝]`은 끝 번호를 포함하지 않아요. `in`은 리스트 안에 값이 있는지 True/False로 알려줍니다.

````python
orders = ["아메리카노", "카페라테", "아메리카노", "치즈케이크", "샌드위치"]
print(orders[0:2])
print(orders[-3:])
````

## S23. 딕셔너리 = 메뉴판
중괄호 `{}`에 `키: 값`을 나열해요. 리스트는 번호로, 딕셔너리는 **이름(키)**으로 꺼냅니다.
이 `menu`는 S24, S25, S27에서 계속 써요.

````python
menu = {
    "아메리카노": 2000,
    "카페라테": 3000,
    "치즈케이크": 4500,
    "샌드위치": 5500
}
print(menu["카페라테"])
print(menu["치즈케이크"])
````

## S24. 메뉴판 고치기: 추가 · 수정 · 삭제
없는 키에 넣으면 **추가**, 있는 키에 넣으면 **수정**, 삭제는 `del`입니다.

````python
menu["바닐라라테"] = 3500
menu["카페라테"] = 3200
del menu["샌드위치"]
print(menu)
````

## S25. 메뉴판에 없는 메뉴를 찾으면?
아래 첫 번째 셀은 **일부러 에러를 내는 셀**입니다. 키가 있는지 모를 땐 `in`으로 확인하거나 `.get()`을 써요.

````python
print(menu["녹차라테"])   # KeyError!
````

````python
print("녹차라테" in menu)
print(menu.get("녹차라테", 0))
````

## S27. 결제 금액은 이렇게 계산된다
괄호가 겹치면 **안쪽부터** 계산해요. `menu[cart[0]]` → `menu["아메리카노"]` → `2000`

````python
cart = ["아메리카노", "치즈케이크"]
print(cart[0])
print(menu["아메리카노"])
print(menu[cart[0]])
````

````python
first = menu[cart[0]]
second = menu[cart[1]]
total = first + second
print(f"결제 금액: {total}원")
````

## S29. 복습 미션: 내 가게 키오스크 만들기
메뉴와 가격을 내 가게로 바꿔 보세요. 아래처럼 나오면 성공!
```
담은 메뉴: 2개
결제 금액: 7000원
```

````python
my_menu = {"떡볶이": 4000, "김밥": 3000}
cart = ["김밥"]
cart.append("떡볶이")

total = my_menu[cart[0]] + my_menu[cart[1]]
print(f"담은 메뉴: {len(cart)}개")
print(f"결제 금액: {total}원")
````
