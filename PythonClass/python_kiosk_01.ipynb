## S23. 딕셔너리 = 메뉴판
이 `menu`는 S24, S25, S27에서 계속 씁니다.

````python
menu = {
    "아메리카노": 2000,
    "카페라테": 3000,
    "치즈케이크": 4500,
    "샌드위치": 5500
}
print(menu["카페라테"])
print(menu["치즈케이크"])
````

## S24. 메뉴판 고치기: 추가 · 수정 · 삭제

````python
menu["바닐라라테"] = 3500
menu["카페라테"] = 3200
del menu["샌드위치"]
print(menu)
````

## S25. 메뉴판에 없는 메뉴를 찾으면?
아래 첫 번째 셀은 **일부러 에러를 내는 셀**입니다.

````python
print(menu["녹차라테"])   # KeyError!
````

````python
print("녹차라테" in menu)
print(menu.get("녹차라테", 0))
````

## S27. 결제 금액은 이렇게 계산된다

````python
cart = ["아메리카노", "치즈케이크"]
print(cart[0])
print(menu["아메리카노"])
print(menu[cart[0]])
````

## S29. 복습 미션: 내 가게 키오스크 만들기
메뉴와 가격을 내 가게로 바꿔 보세요. 아래처럼 나오면 성공!
```
담은 메뉴: 2개
결제 금액: 7000원
```

````python
my_menu = {"떡볶이": 4000, "김밥": 3000}
cart = ["김밥"]
cart.append("떡볶이")

total = my_menu[cart[0]] + my_menu[cart[1]]
print(f"담은 메뉴: {len(cart)}개")
print(f"결제 금액: {total}원")
````
