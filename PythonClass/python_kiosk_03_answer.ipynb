## 실습 준비 (S6): 신메뉴와 비건 정보
신메뉴 2개(오트라테, 과일컵)와 메뉴별 비건 여부(`vegan`)를 추가했어요. 지난 시간의 `calc_total` 함수도 들어 있습니다. <b>가장 먼저 실행</b>하세요.

````python
menu  = {"아메리카노": 2000, "카페라테": 3000, "치즈케이크": 4500,
         "샌드위치": 5500, "오트라테": 3500, "과일컵": 4000}
stock = {"아메리카노": 10, "카페라테": 5, "치즈케이크": 0,
         "샌드위치": 2, "오트라테": 4, "과일컵": 3}
vegan = {"아메리카노": True, "카페라테": False, "치즈케이크": False,
         "샌드위치": False, "오트라테": True, "과일컵": True}

# 2강-2에서 만든 결제 금액 함수
def calc_total(cart):
    total = 0
    for item in cart:
        total = total + menu[item]
    return total

print(menu)
````

## S8. 지금까지 배운 방법으로 고르기
빈 리스트 → 하나씩 꺼내기 → 조건 확인 → 담기. 네 줄이 필요해요.

````python
cheap = []
for name in menu:
    if menu[name] <= 5000:
        cheap.append(name)
print(cheap)
````

## S9. 리스트 컴프리헨션: 한 줄로
`[담을 값 for 변수 in 대상 if 조건]` 대괄호 안에 "무엇을 · 어디서 · 어떤 조건으로"를 씁니다.

````python
cheap = [name for name in menu if menu[name] <= 5000]
print(cheap)
````

## S10. 비건 메뉴만 골라내기
`vegan[name]`은 이미 True/False라서 그대로 조건이 돼요.

````python
vegan_list = [name for name in menu if vegan[name]]
print(vegan_list)
````

## S11. 골라내면서 모양까지 바꾸기
맨 앞 "담을 값" 자리에 f-string을 쓰면 값을 바꿔서 담을 수 있어요. 재고가 0인 메뉴는 빠집니다.

````python
screen = [f"{name} {menu[name]}원"
          for name in menu if stock[name] > 0]
for line in screen:
    print(line)
````

## S12. [선택] 딕셔너리 컴프리헨션
중괄호 `{}` 안에 `키: 값`을 쓰면 새 딕셔너리가 만들어져요.

````python
event = {name: menu[name] - 500 for name in menu}
print(event["아메리카노"], event["샌드위치"])
````

## S14. lambda: 한 줄짜리 함수
`lambda 매개변수: 돌려줄 값` return은 쓰지 않아요. 짧고 한 번 쓰고 말 함수에 씁니다.

````python
def discount(price):
    return price - 500

discount2 = lambda price: price - 500

print(discount(3000), discount2(3000))
````

## S15. sorted + lambda: 가격순으로 정렬
`key=`에 "무엇을 기준으로 정렬할지" 함수를 넘겨요. 비싼 순서로 바꾸려면 `reverse=True`를 붙입니다.

````python
by_price = sorted(menu, key=lambda name: menu[name])
print(by_price)
````

## S16. [선택] filter + lambda
결과는 컴프리헨션과 같아요.

````python
cheap2 = list(filter(lambda name: menu[name] <= 5000, menu))
print(cheap2)
````

## S18. 영수증 출력 함수
장바구니와 합계를 받아서 영수증을 출력해요. 이 `print_receipt`는 S19~S21에서 씁니다.

````python
def print_receipt(cart, total):
    print("===== 영수증 =====")
    for item in cart:
        print(f"{item}  {menu[item]}원")
    print(f"합계: {total}원")

cart = ["아메리카노", "오트라테"]
print_receipt(cart, calc_total(cart))
````

## S19. 콜백: 결제가 끝나면 할 일을 넘겨주기
함수를 매개변수로 넘겨서, 일이 끝난 뒤 <b>나중에</b> 부르게 하는 것을 콜백이라고 해요.

````python
def pay(cart, on_done):
    total = calc_total(cart)
    print(f"{total}원 결제 완료")
    on_done(cart, total)

pay(cart, print_receipt)
````

## S20. 할 일 바꿔 끼우기
`pay`는 그대로 두고, 넘겨주는 함수만 바꿔요. 한 번만 쓸 함수는 람다로 바로 넘깁니다.

````python
def send_alarm(cart, total):
    print(f"[알림] 주문 {len(cart)}건이 접수되었습니다")

pay(cart, send_alarm)
````

````python
pay(cart, lambda c, t: print(f"포인트 {int(t / 100)}점 적립"))
````

## S21. 주의: 괄호를 붙이면 안 돼요
아래 셀은 <b>일부러 에러를 내는 셀</b>입니다. 함수를 넘길 때는 이름만 씁니다. 괄호를 붙이면 "지금 바로 실행"이 돼요.

````python
pay(cart, print_receipt())   # TypeError!
````

## S24. 복습 미션: 내 분식집 메뉴 필터와 결제
메뉴를 내 가게로 바꿔 보세요. 아래처럼 나오면 성공!
```
['김밥', '라면']
['김밥', '라면', '떡볶이']
결제 금액: 6500원
```

````python
my_menu = {"떡볶이": 4000, "김밥": 3000, "라면": 3500}

cheap = [name for name in my_menu if my_menu[name] <= 3500]
print(cheap)
print(sorted(my_menu, key=lambda name: my_menu[name]))

def pay_my(cart, on_done):
    total = 0
    for item in cart:
        total = total + my_menu[item]
    on_done(total)

pay_my(["김밥", "라면"], lambda t: print(f"결제 금액: {t}원"))
````
