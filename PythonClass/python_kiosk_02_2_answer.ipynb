## 실습 준비: 메뉴판과 재고
이번 시간 내내 쓰는 데이터예요. <b>가장 먼저 실행</b>하세요. 주문 처리(S17)를 다시 해 보고 싶을 때도 이 셀부터 다시 실행하면 재고가 처음으로 돌아갑니다.

````python
menu  = {"아메리카노": 2000, "카페라테": 3000, "치즈케이크": 4500, "샌드위치": 5500}
stock = {"아메리카노": 10,   "카페라테": 5,    "치즈케이크": 0,    "샌드위치": 2}
print(menu)
print(stock)
````

## S7. def: 함수 만들고 부르기
`def 이름():`으로 함수를 <b>정의</b>(버튼 만들기)하고, `이름()`으로 <b>호출</b>(버튼 누르기)해요. 정의만 하면 아무 일도 일어나지 않습니다.

````python
def greet():
    print("어서 오세요! 파이썬 카페입니다.")
````

````python
greet()
greet()
````

## S8. 매개변수: 값을 넘겨주기
괄호 안의 `store`를 <b>매개변수</b>라고 해요. 호출할 때 넣은 값이 `store`에 들어갑니다.

````python
def greet(store):
    print(f"어서 오세요! {store}입니다.")

greet("파이썬 카페")
greet("행복 분식")
````

## S9. 매개변수 두 개: 메뉴와 수량
매개변수는 쉼표로 여러 개 둘 수 있어요. 호출할 때 <b>같은 순서</b>로 값을 넣습니다.

````python
def show_item(name, qty):
    price = menu[name] * qty
    print(f"{name} {qty}개 = {price}원")

show_item("아메리카노", 2)
show_item("샌드위치", 1)
````

## S11. return: 계산해서 돌려주기
`return`은 함수가 계산한 값을 부른 곳으로 돌려줘요. 2강-1의 합계 코드를 함수로 만들었습니다.
이 `calc_total`은 S12, S17에서도 써요.

````python
def calc_total(cart):
    total = 0
    for item in cart:
        total = total + menu[item]
    return total

a = calc_total(["아메리카노", "카페라테"])
b = calc_total(["카페라테", "샌드위치", "아메리카노"])
print(a, b)
````

## S12. 돌려받은 값으로 거스름돈 계산
return으로 받은 값은 변수에 넣어 다음 계산에 바로 쓸 수 있어요.

````python
cart = ["카페라테", "샌드위치"]
total = calc_total(cart)
change = 10000 - total
print(f"결제 금액: {total}원")
print(f"거스름돈: {change}원")
````

## S13. print와 return은 달라요
`print`는 화면에 보여주기만 하고, return이 없는 함수는 `None`(아무것도 없음)을 돌려줘요.

````python
def show_total(cart):
    print(calc_total(cart))

t = show_total(["아메리카노"])
print(t)
````

아래 셀은 <b>일부러 에러를 내는 셀</b>입니다. (None으로는 계산할 수 없어요)

````python
print(10000 - t)   # TypeError!
````

## S14. True / False를 돌려주는 함수
`return`을 만나면 그 자리에서 함수가 끝나요. 이 `can_order`는 S16에서도 씁니다.

````python
def can_order(name):
    if name not in menu:
        return False
    if stock[name] == 0:
        return False
    return True

print(can_order("아메리카노"))
print(can_order("치즈케이크"))
print(can_order("녹차라테"))
````

## S16. 장바구니 담기 함수
함수 안에서 다른 함수(`can_order`)를 부를 수 있어요. `not`은 True/False를 뒤집습니다. (정의만 하는 셀이라 출력은 없어요)

````python
def add_to_cart(name, qty):
    if not can_order(name):
        print(f"{name}: 주문할 수 없습니다")
    elif stock[name] < qty:
        print(f"{name}: 재고가 {stock[name]}개뿐입니다")
    else:
        for i in range(qty):
            cart.append(name)
        stock[name] = stock[name] - qty
        print(f"{name} {qty}개 담기 완료")
````

## S17. 주문 처리 프로그램 완성
오늘 만든 함수 세 개가 함께 일해요.
> 다시 실행하면 재고가 계속 줄어들어요. 처음 결과로 돌아가려면 맨 위 '실습 준비' 셀부터 다시 실행하세요.

````python
cart = []
add_to_cart("아메리카노", 2)
add_to_cart("치즈케이크", 1)
add_to_cart("샌드위치", 3)
add_to_cart("카페라테", 1)
print(cart)
print(f"결제 금액: {calc_total(cart)}원")
````

## S21. 복습 미션: 내 분식집 계산 함수
메뉴와 가격을 내 가게로 바꿔 보세요. 아래처럼 나오면 성공!
```
6500
10000
```

````python
my_menu = {"떡볶이": 4000, "김밥": 3000, "라면": 3500}

def calc_my_total(cart):
    total = 0
    for item in cart:
        total = total + my_menu[item]
    return total

print(calc_my_total(["김밥", "라면"]))
print(calc_my_total(["떡볶이", "김밥", "김밥"]))
````
