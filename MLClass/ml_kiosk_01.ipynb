## S33. [선택] GPU는 얼마나 빠를까?
코드는 지금 몰라도 괜찮아요. CPU와 GPU의 시간만 비교해 보세요. (숫자는 실행할 때마다 달라요)

```python
import time
a = torch.rand(4000, 4000)       # 숫자 1,600만 개짜리 표
start = time.time()
b = a @ a                        # 큰 곱셈 (CPU)
print("CPU:", round(time.time() - start, 3), "초")

a = a.to("cuda")                 # 표를 GPU로 옮기기
b = a @ a; torch.cuda.synchronize()    # 준비 운동
start = time.time()
b = a @ a; torch.cuda.synchronize()    # 계산이 끝날 때까지 기다리기
print("GPU:", round(time.time() - start, 3), "초")
```
