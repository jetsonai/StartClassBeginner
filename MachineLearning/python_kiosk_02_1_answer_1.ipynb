## 실습 준비: 메뉴판과 재고
이번 시간 내내 쓰는 데이터예요. <b>가장 먼저 실행</b>하세요.

````python
menu  = {"아메리카노": 2000, "카페라테": 3000, "치즈케이크": 4500, "샌드위치": 5500}
stock = {"아메리카노": 10,   "카페라테": 5,    "치즈케이크": 0,    "샌드위치": 2}
print(menu)
print(stock)
````

## S7. 비교 연산자
`== != > < >= <=`로 비교하면 결과가 True/False로 나와요. `=`는 대입, `==`는 비교입니다.

````python
stock_now = 3
print(stock_now == 0)
print(stock_now > 0)

price = 4500
print(price <= 5000)
````

## S8. if: 품절이면 막기
`if 조건:` 끝에 <b>콜론</b>, 다음 줄은 <b>4칸 들여쓰기</b>. 조건이 True일 때만 들여쓴 줄이 실행돼요. `stock_cake`를 5로 바꿔서도 실행해 보세요.

````python
stock_cake = 0
if stock_cake == 0:
    print("품절된 메뉴입니다")
````

## S9. 들여쓰기가 범위를 정한다
들여쓴 줄은 조건이 True일 때만, 들여쓰지 않은 줄은 항상 실행됩니다.

````python
if stock_cake == 0:
    print("품절된 메뉴입니다")
print("다른 메뉴를 골라 주세요")
````

아래 셀은 <b>일부러 에러를 내는 셀</b>입니다. (들여쓰기를 빼먹은 경우)

````python
if stock_cake == 0:
print("품절된 메뉴입니다")   # IndentationError!
````

## S10. if-else: 막거나, 담거나
`else`는 if 조건이 False일 때 실행돼요. 둘 중 하나는 반드시 실행됩니다. `"치즈케이크"`를 `"아메리카노"`로 바꿔서도 실행해 보세요.

````python
if stock["치즈케이크"] == 0:
    print("품절된 메뉴입니다")
else:
    print("장바구니에 담았습니다")
````

## S11. elif: 품절 임박 알림
위에서부터 확인해서 <b>처음 True가 되는 블록 하나만</b> 실행돼요. `stock_latte`를 0, 2, 7로 바꿔 가며 실행해 보세요.

````python
stock_latte = 2
if stock_latte == 0:
    print("품절")
elif stock_latte <= 3:
    print(f"품절 임박! 남은 수량 {stock_latte}개")
else:
    print("주문 가능")
````

## S12. in과 함께: 없는 메뉴 거르기
`in`은 True/False를 돌려주니까 그대로 if 조건이 돼요.

````python
order = "녹차라테"
if order in menu:
    print(f"{order}: {menu[order]}원")
else:
    print("판매하지 않는 메뉴입니다")
````

## S13. and · or: 쿠폰 할인
`and`는 둘 다 True일 때, `or`는 하나만 True여도, `not`은 True와 False를 뒤집어요.

````python
total = 12000
has_coupon = True
if total >= 10000 and has_coupon:
    total = total - 1000
print(total)
````

## S14. 주문 담기 로직 완성
없는 메뉴 → 품절 → 담기 순서로 확인해요. `order`를 "녹차라테", "치즈케이크", "샌드위치"로 바꿔 가며 실행해 보세요.

````python
cart = []
order = "샌드위치"
if order not in menu:
    print("판매하지 않는 메뉴입니다")
elif stock[order] == 0:
    print("품절된 메뉴입니다")
else:
    cart.append(order)
    stock[order] = stock[order] - 1
    print(f"{order} 담았습니다")
    print(f"남은 수량: {stock[order]}개")
````

## S16. for: 하나씩 꺼내기
`for 변수 in 리스트:` 리스트의 값을 처음부터 하나씩 변수에 넣고, 들여쓴 줄을 반복해요.

````python
cart = ["아메리카노", "카페라테", "샌드위치"]
for item in cart:
    print(item)
````

## S17. for로 결제 금액 구하기
`total = 0`에서 시작해 반복할 때마다 가격을 더해요. 메뉴가 몇 개든 코드는 그대로입니다.

````python
total = 0
for item in cart:
    total = total + menu[item]
print(f"결제 금액: {total}원")
````

## S18. for + if: 키오스크 메뉴 화면
딕셔너리에 `for`를 쓰면 <b>키</b>(메뉴 이름)가 하나씩 나와요.

````python
for name in menu:
    if stock[name] == 0:
        print(f"{name}  (품절)")
    else:
        print(f"{name}  {menu[name]}원")
````

## S19. range: 주문 번호 만들기
`range(시작, 끝)`은 시작부터 <b>끝 바로 앞까지</b>예요. 1강 슬라이싱과 같은 규칙입니다.

````python
for num in range(101, 104):
    print(f"주문번호 {num}번 준비 완료")
````

## S20. while: 재고가 남아 있는 동안
`while`은 조건이 True인 동안 반복해요. 멈추지 않으면 셀 왼쪽의 정지 버튼을 누르세요.

````python
cake = 3
while cake > 0:
    cake = cake - 1
    print(f"치즈케이크 판매! 남은 수량 {cake}개")
print("품절되었습니다")
````

## S23. 복습 미션: 내 분식집 메뉴 화면
메뉴와 재고를 내 가게로 바꿔 보세요. 아래처럼 나오면 성공!
```
떡볶이: 품절
김밥: 3000원
라면: 3500원
결제 금액: 9500원
```

````python
my_menu  = {"떡볶이": 4000, "김밥": 3000, "라면": 3500}
my_stock = {"떡볶이": 0, "김밥": 5, "라면": 3}

for name in my_menu:
    if my_stock[name] == 0:
        print(f"{name}: 품절")
    else:
        print(f"{name}: {my_menu[name]}원")

my_cart = ["김밥", "라면", "김밥"]
total = 0
for item in my_cart:
    total = total + my_menu[item]
print(f"결제 금액: {total}원")
````
